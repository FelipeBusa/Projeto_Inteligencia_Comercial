```python
# Instalar as bibliotecas utilizadas
# pip install kagglehub pandas -q

# Importar as bibliotecas utilizadas
import kagglehub
from pathlib import Path
import shutil
```

```python
# Realizar o download do dataset
caminho_download = kagglehub.dataset_download(
    "olistbr/brazilian-ecommerce"
)

print("Arquivos baixados:", caminho_download)
```

```python
# Criar o diretório RAW
pasta_raw = Path("data/raw")
pasta_raw.mkdir(parents=True, exist_ok=True)
```

```python
# Copiar os arquivos originais
for arquivo in Path(caminho_download).glob("*.csv"):
    shutil.copy2(
        arquivo,
        pasta_raw / arquivo.name
    )

print("Arquivos armazenados na camada RAW.")
```

```python
# Conferir o resultado
arquivos = list(pasta_raw.glob("*.csv"))
print(f"Total de arquivos: {len(arquivos)}")

for arquivo in sorted(arquivos):
    print(arquivo.name)
```
